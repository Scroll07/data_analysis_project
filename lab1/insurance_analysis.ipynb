```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Загружаем данные
df = pd.read_csv('insurance.csv')
display(df.head())
```

```python
print("=== Типы данных до преобразования ===")
print(df.dtypes)
print("\n=== Проверка пропусков ===")
print(df.isnull().sum())
```

```python
# Преобразуем текстовые колонки в более оптимальный тип 'category'
df['sex'] = df['sex'].astype('category')
df['smoker'] = df['smoker'].astype('category')
df['region'] = df['region'].astype('category')

print("=== Типы данных после преобразования ===")
print(df.dtypes)
```

```python
plt.figure(figsize=(15, 5))

# 1. Гистограмма возраста
plt.subplot(1, 3, 1)
sns.histplot(df['age'], bins=20, kde=True, color='skyblue')
plt.title('Распределение возраста')

# 2. Влияние курения на расходы
plt.subplot(1, 3, 2)
sns.boxplot(x='smoker', y='charges', data=df, palette='Set2')
plt.title('Расходы: курящие vs некурящие')

# 3. Индекс массы тела и расходы
plt.subplot(1, 3, 3)
sns.scatterplot(x='bmi', y='charges', hue='smoker', data=df)
plt.title('Зависимость расходов от ИМТ')

plt.tight_layout()
plt.show()
```
